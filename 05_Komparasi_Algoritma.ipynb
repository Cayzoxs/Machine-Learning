# 1. Persiapan Data dan Inisialisasi Model

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report

url = "https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv"
df = pd.read_csv(url)
X_scaled = StandardScaler().fit_transform(df.drop('Outcome', axis=1))
X_train, X_test, y_train, y_test = train_test_split(X_scaled, df['Outcome'], test_size=0.2, random_state=42)

knn = KNeighborsClassifier(n_neighbors=5)
dt = DecisionTreeClassifier(max_depth=3, random_state=42)
```

# 2. Pengujian Komparatif (KNN vs Decision Tree)

```python
knn.fit(X_train, y_train)
dt.fit(X_train, y_train)

print("=== Laporan Performa KNN ===")
print(classification_report(y_test, knn.predict(X_test)))

print("\n=== Laporan Performa Decision Tree ===")
print(classification_report(y_test, dt.predict(X_test)))
```
