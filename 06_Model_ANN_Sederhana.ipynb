# 1. Pra-pemrosesan dan Pelatihan Artificial Neural Network

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

url = "https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv"
df = pd.read_csv(url)

X_scaled = StandardScaler().fit_transform(df.drop('Outcome', axis=1))
X_train, X_test, y_train, y_test = train_test_split(X_scaled, df['Outcome'], test_size=0.2, random_state=42)

# Menggunakan 2 hidden layer
ann_model = MLPClassifier(hidden_layer_sizes=(8, 4), max_iter=1000, random_state=42)
ann_model.fit(X_train, y_train)

print(f"Akurasi Model Artificial Neural Network: {accuracy_score(y_test, ann_model.predict(X_test)) * 100:.2f}%")
```
