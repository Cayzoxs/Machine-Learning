# 1. Mengimpor Library dan Memuat Dataset

```python
# 1. Mengimpor library yang dibutuhkan
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Memuat dataset dari internet
url = "https://raw.githubusercontent.com/plotly/datasets/master/diabetes.csv"
df = pd.read_csv(url)

# 3. Melihat 5 baris pertama data
print("--- 5 Baris Pertama Dataset ---")
display(df.head())

# 4. Mengecek informasi kolom dan nilai kosong
print("\n--- Informasi Dataset ---")
print(df.info())
print("\nJumlah Data Kosong:")
print(df.isnull().sum())

# 5. Visualisasi perbandingan target (0 = Tidak Diabetes, 1 = Diabetes)
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Outcome', palette='Set2')
plt.title("Perbandingan Pasien Diabetes (1) dan Tidak Diabetes (0)")
plt.show()
```
